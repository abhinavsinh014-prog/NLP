```python
import nltk

nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize

resume = {
    "name": "John Doe",
    "role": "Software Engineer",
    "skills": ["Python", "JavaScript", "React", "Django"],
    "experience": ["2 years"]
}

text = f"""
Name: {resume['name']}
Role: {resume['role']}
Skills: {', '.join(resume['skills'])}
Experience: {resume['experience'][0]}
"""

doc = sent_tokenize(text)

print(doc)
```

```python
for sent in doc:
  print(sent)
```

```python
word = nltk.word_tokenize(text)
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
stemmer= PorterStemmer()
import nltk
nltk.download('stopwords')
```

```python
for i in range(len(word)):
  words=nltk.word_tokenize(word[i])
  words=[stemmer.stem(word) for word in words if word not in set(word)]
  word[i] = ' '.join(words)
```

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmi = WordNetLemmatizer()

for i in range(len(word)):
  words=nltk.word_tokenize(word[i])
  words=[lemmi.lemmatize(word) for word in words if word not in set(word)]
  word[i] = ' '.join(words)
```
