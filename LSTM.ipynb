```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    'the ball of the century',
    'greatest of all time',
    'head of the table',
    'the peoples champ',
    'rest in peace'
]
```

```python
sent
```

```python
voc_size = 10000
```

```python
one = [one_hot(words, voc_size) for words in sent]
print(one)
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length = 8
pre_pad = pad_sequences(one, padding='pre', maxlen=sent_length)
```

```python
print(pre_pad)
```

```python
post_pad = pad_sequences(one, padding='post', maxlen=sent_length)
```

```python
print(post_pad)
```

```python
dim = 10
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Input

model = Sequential([
    Input(shape=(sent_length,)),
    Embedding(input_dim=voc_size, output_dim=10)
])

model.compile(optimizer='adam', loss='mse')

model.summary()
```
