```python
words = ["eating","eaten","eats","writing","writes","programming","programs","history","going","goes","go"]
```

```python
from nltk import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
  print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem("understanding")
```

```python
stemming.stem("congrulation")
```

```python
from nltk import RegexpStemmer
```

```python
reg = RegexpStemmer('ing$|s$|es$|able$|',min=4)
```

```python
reg.stem('AuraFarming')
```

```python
reg.stem('ingAuraFarming')
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball = SnowballStemmer('english')
```

```python
for word in words:
  print(word+"---->"+snowball.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportly")
```

```python
snowball.stem("fairly"),snowball.stem("sportly")
```
