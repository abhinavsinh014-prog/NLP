```python
import pandas as pd

msg = pd.read_csv('/content/SMSSpamCollection (1).txt',sep='\t',names=['label','message'])

```

```python
msg.shape
```

```python
msg['label'].loc[451],msg['message'].loc[451]
```

```python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0,len(msg)):
  review = re.sub('[^a-zA-Z]',' ',msg['message'][i])
  review = review.lower()
  review = review.split()

  review = [ps.stem(word) for word in review if not word  in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv  = CountVectorizer(max_features=2500,binary = True,ngram_range=(2,2))
X = cv.fit_transform(corpus).toarray()
```

```python
X.shape
```

```python
y = pd.get_dummies(msg['label'])
y = y.iloc[:,1].values
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model = MultinomialNB().fit(X_train,y_train)
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
score = accuracy_score(y_test,y_pred)
```

```python
score
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
score2 = accuracy_score(y_test,y_pred)
```

```python
score2
```

```python
from sklearn.ensemble import RandomForestClassifier
spam_detect_model = RandomForestClassifier().fit(X_train,y_train)
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
```
