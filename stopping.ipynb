```python
para = """I am an AI/ML-focused developer passionate about building practical intelligent applications and continuously expanding my knowledge in Artificial Intelligence.

My current foundation is built around Python, Streamlit, Machine Learning, and Deep Learning. I enjoy taking theoretical concepts and turning them into functional applications that solve practical problems.

I am currently gaining professional experience through an AI Internship at Smarted, where I am developing practical exposure to AI-focused development and real-world problem solving.

My next major learning goals are Large Language Models (LLMs), Retrieval-Augmented Generation (RAG), and Agentic AI."""
```

```python
print(para)
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.stem import PorterStemmer
stemmer= PorterStemmer()
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwards = set(stopwords.words('english'))
```

```python
stopwards
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
sentence = nltk.sent_tokenize(para)
```

```python
type(sentence)
```

```python
for i in range(len(sentence)):
  words=nltk.word_tokenize(sentence[i])
  words=[stemmer.stem(word) for word in words if word not in set(stopwards)]
  sentence[i] = ' '.join(words)
```

```python
sentence
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball = SnowballStemmer('english')
```

```python
for i in range(len(sentence)):
  words=nltk.word_tokenize(sentence[i])
  words=[snowball.stem(word) for word in words if word not in set(stopwards)]
  sentence[i] = ' '.join(words)
```

```python
sentence
```

```python
import nltk
nltk.download('wordnet')
```

```python
#lemmitasation
from nltk.stem import WordNetLemmatizer
lemmi = WordNetLemmatizer()

for i in range(len(sentence)):
  words=nltk.word_tokenize(sentence[i])
  words=[lemmi.lemmatize(word) for word in words if word not in set(stopwards)]
  sentence[i] = ' '.join(words)
```
